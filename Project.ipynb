```python
class Expense:
    def __init__(self,amount,category,date,payment_method):
        self.amount=amount
        self.category=category
        self.date=date
        self.payment_method=payment_method
    def __str__(self):
        return f"Amount: {self.amount}, Category: {self.category}, Date: {self.date}, Payment_method: {self.payment_method}"
e1=Expense(500,"shopping","14-Aug","cash") 
print(e1)
```

```python
class ExpenseTracker:
    def __init__(self):
        self.expenses=[]
    def add_expense(self,amount,category,date,payment_method):
        new_expense=Expense(amount,category,date,payment_method)
        self.expenses.append(new_expense)
    def get_total(self):
        total = 0
        for expense in self.expenses:
             total += expense.amount
        return total
    def get_category_total(self,category):
        total=0
        for expense in self.expenses:
            if expense.category.lower()==category.lower():
                total+=expense.amount
        return total
    def save_to_file(self,filename):
        with open(filename,"w") as file:
           for expense in self.expenses:
                line = f"{expense.amount},{expense.category},{expense.date},{expense.payment_method}\n"
                file.write(line)
    def load_from_file(self, filename):
        self.expenses = []
        with open(filename, "r") as file:
             for line in file:
                 parts = line.strip().split(",")
                 amount = float(parts[0])
                 category = parts[1]
                 date = parts[2]
                 payment_method = parts[3]
                 new_expense = Expense(amount, category, date, payment_method)
                 self.expenses.append(new_expense)
```

```python
new_tracker = ExpenseTracker()
new_tracker.load_from_file("expenses.txt")

for expense in new_tracker.expenses:
    print(expense)
```

```python

tracker=ExpenseTracker()
tracker.add_expense(2000,"Bills","14-2-23","Cash")
tracker.add_expense(4000,"shopping","14-2-23","Cash")
tracker.add_expense(2000,"food","15-2-23","Cash")
print(tracker.expenses)
```

```python
for expense in tracker.expenses:
    print(expense)
```

```python
print("Total expense:",tracker.get_total())
```

```python
print("Food total:",tracker.get_category_total("food"))
print("Bills total:",tracker.get_category_total("Bills"))
```

```python
tracker.save_to_file("expenses.txt")
```
